### 🧠 RISC-V 5-Stage Pipelined CPU Architecture

---

## 1. Introduction

This project implements a **32-bit RISC-V (RV32I-like) pipelined processor** in Verilog HDL.
The design follows a classical **5-stage pipeline architecture** with hazard handling and modular RTL components.

The CPU supports:

* Arithmetic operations
* Logical operations
* Load/store instructions
* Branch instructions
* Immediate-based operations

---

## 2. CPU Architecture Overview

The processor is divided into five pipeline stages:

```
IF  →  ID  →  EX  →  MEM  →  WB
```

### Pipeline Registers:

* IF/ID
* ID/EX
* EX/MEM
* MEM/WB

Each stage operates concurrently, improving instruction throughput.

---

## 3. Block Diagram Description

The CPU consists of the following major components:

### Instruction Fetch (IF)

* Program Counter (PC)
* Instruction Memory
* PC update logic

### Instruction Decode (ID)

* Control Unit
* Register File
* Immediate Generator

### Execute (EX)

* ALU
* ALU Control Unit
* Branch Unit

### Memory (MEM)

* Data Memory

### Write Back (WB)

* Register File update logic

---

## 4. Instruction Flow

Each instruction passes through the pipeline as follows:

### IF Stage

* Fetch instruction from memory using PC

### ID Stage

* Decode instruction opcode
* Read source registers
* Generate immediate value
* Generate control signals

### EX Stage

* Perform ALU operation
* Compute branch condition

### MEM Stage

* Perform load/store operations

### WB Stage

* Write result back to register file

## 6. Control Unit

The control unit generates signals:

| Signal   | Function                      |
| -------- | ----------------------------- |
| RegWrite | Register write enable         |
| MemRead  | Memory read enable            |
| MemWrite | Memory write enable           |
| MemToReg | Select memory or ALU result   |
| ALUSrc   | Immediate or register operand |
| Branch   | Branch enable                 |
| ALUOp    | ALU operation control         |

* I-Type

---

## 9. Hazard Handling (Design Level)

### Forwarding Unit

* Resolves data hazards
* Forwards results from MEM/WB stages

### Hazard Detection Unit

* Detects load-use hazards
* Introduces pipeline stalls when required

---

## 10. Branch Handling

Branch decisions are computed in EX stage using:

* ALU comparison
* Branch condition logic

If branch is taken:

```
PC = PC + Immediate
```

Else:

```
PC = PC + 4
```

---

## 11. Memory System

### Instruction Memory

* Stores program.hex instructions
* Read-only during execution

### Data Memory

* Supports load/store operations
* Word-addressable memory

---

## 12. Pipeline Summary

| Stage | Function               |
| ----- | ---------------------- |
| IF    | Fetch instruction      |
| ID    | Decode + register read |
| EX    | ALU + branch decision  |
| MEM   | Memory access          |
| WB    | Write result           |

---

## 13. Example Program Execution

Example program:

```
addi x1, x0, 5
addi x2, x0, 10
add  x3, x1, x2
sw   x3, 0(x0)
```

### Final result:

* x1 = 5
* x2 = 10
* x3 = 15
* memory[0] = 15
